```python
grocery = {
    "rice": {"price": 50, "stock": 100},
    "sugar": {"price": 45, "stock": 80},
    "oil": {"price": 150, "stock": 40},
    "milk": {"price": 30, "stock": 60}
}

item = input("Enter product name: ").lower()

if item in grocery:
    quantity = int(input("Enter quantity: "))

    if quantity <= grocery[item]["stock"]:
        price = grocery[item]["price"]
        total_bill = price * quantity

        grocery[item]["stock"] -= quantity

        print("Product:", item)
        print("Quantity:", quantity)
        print("Price per unit: ₹", price)
        print("Total Bill: ₹", total_bill)
        print("Remaining Stock:", grocery[item]["stock"])

    else:
        print("Not enough stock!")

else:
    print("Product not found!")
```

```python
grocery = {
    "rice": {"price": 50, "stock": 100},
    "sugar": {"price": 45, "stock": 80},
    "oil": {"price": 150, "stock": 40},
    "milk": {"price": 30, "stock": 60}
}

total_bill = 0

while True:
    item = input("Enter product name (or done): ").lower()

    if item == "done":
        break

    if item not in grocery:
        print("Product not found!")
        continue

    quantity = int(input("Enter quantity: "))

    if quantity <= 0:
        print("Quantity must be greater than zero!")
        continue

    if quantity > grocery[item]["stock"]:
        print("Not enough stock!")
        continue

    price = grocery[item]["price"]
    bill = price * quantity

    grocery[item]["stock"] -= quantity
    total_bill += bill

    print("Item Bill: ₹", bill)
    print("Remaining Stock:", grocery[item]["stock"])

print("Final Total Bill: ₹", total_bill)
```

```python
grocery = {
    "rice": {"price": 50, "stock": 100},
    "sugar": {"price": 45, "stock": 80},
    "oil": {"price": 150, "stock": 40},
    "milk": {"price": 30, "stock": 60}
}

cart = []
total_bill = 0

while True:
    item = input("Enter product name (or done): ").lower()

    if item == "done":
        break

    if item not in grocery:
        print("Product not found!")
        continue

    quantity = int(input("Enter quantity: "))

    if quantity <= 0:
        print("Quantity must be greater than zero!")
        continue

    if quantity > grocery[item]["stock"]:
        print("Not enough stock!")
        continue

    price = grocery[item]["price"]
    bill = price * quantity

    grocery[item]["stock"] -= quantity
    total_bill += bill

    cart.append({
        "product": item,
        "quantity": quantity,
        "price": price,
        "total": bill
    })

    print("Item added! Bill: ₹", bill)

print("\n===== GROCERY SHOP RECEIPT =====")

for product in cart:
    print(
        product["product"],
        "| Qty:", product["quantity"],
        "| Price: ₹", product["price"],
        "| Total: ₹", product["total"]
    )

print("--------------------------------")
print("Final Total Bill: ₹", total_bill)
print("Thank you for shopping!")
```
